```python
# ============================================
# PUSH PROJECT TO GITHUB FROM COLAB
# Canada-Brazil Trade Report
# ============================================

import os
import shutil
import getpass

# ============================================
# 1. MOUNT GOOGLE DRIVE SAFELY
# ============================================
IN_COLAB = False

try:
    from google.colab import drive
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

PROJECT_PATH = "/content/drive/MyDrive/canada-brazil-trade-report"

if IN_COLAB:
    if not os.path.exists("/content/drive/MyDrive"):
        print("Mounting Google Drive...")
        try:
            drive.mount("/content/drive", force_remount=True)
            print("✅ Google Drive mounted.")
        except Exception as e:
            raise RuntimeError(
                "Google Drive mount failed. Restart the Colab runtime, "
                "run the mount cell again, and complete the Google auth popup."
            ) from e
    else:
        print("✅ Google Drive already available.")
else:
    print("⚠️ Not running in Colab. Drive mount skipped.")

# ============================================
# 2. GO TO PROJECT FOLDER
# ============================================
if not os.path.exists(PROJECT_PATH):
    raise FileNotFoundError(
        f"Project folder not found: {PROJECT_PATH}\n"
        "Make sure the folder exists in Google Drive with this exact name:\n"
        "canada-brazil-trade-report"
    )

os.chdir(PROJECT_PATH)
print(f"📂 Working directory: {os.getcwd()}")

# ============================================
# 3. GITHUB SETTINGS
# ============================================
username = "juliocezarcarneiro"
repo_name = "canada-brazil-trade-report"
user_email = "juliocezarcarneiro@outlook.com"
user_name = "juliocezarcarneiro"

token = getpass.getpass("Enter your GitHub token: ")

# ============================================
# 4. PREPARE GITHUB PAGES FILE
# ============================================
os.makedirs("docs", exist_ok=True)

source_html = os.path.join("reports", "trade-report.html")
target_html = os.path.join("docs", "index.html")

if os.path.exists(source_html):
    shutil.copy(source_html, target_html)
    print("✅ Copied reports/trade-report.html → docs/index.html")
else:
    print("⚠️ reports/trade-report.html not found.")
    with open(target_html, "w", encoding="utf-8") as f:
        f.write("""
<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <title>Canada-Brazil Trade Report</title>
</head>
<body>
  <h1>Canada-Brazil Trade Report</h1>
  <p>The HTML report has not been generated yet.</p>
</body>
</html>
""".strip())
    print("✅ Created fallback docs/index.html")

# ============================================
# 5. OPTIONAL .GITIGNORE
# ============================================
gitignore_content = """__pycache__/
*.pyc
.DS_Store
.ipynb_checkpoints/
"""

if not os.path.exists(".gitignore"):
    with open(".gitignore", "w", encoding="utf-8") as f:
        f.write(gitignore_content)
    print("✅ Created .gitignore")

# ============================================
# 6. RESET AND INIT GIT
# ============================================
print("\n🔄 Resetting git repository...")
!rm -rf .git
!git init
!git branch -M main
!git config user.name "{user_name}"
!git config user.email "{user_email}"

# ============================================
# 7. ADD, COMMIT, PUSH
# ============================================
print("\n📦 Adding files...")
!git add .

print("\n📊 Git status:")
!git status

print("\n💾 Creating commit...")
!git commit -m "Add Canada-Brazil trade report project" || echo "No new changes to commit"

print("\n🔗 Setting remote...")
!git remote remove origin 2> /dev/null
!git remote add origin https://{username}:{token}@github.com/{username}/{repo_name}.git

print("\n🚀 Pushing to GitHub...")
!git push -u origin main --force

print("\n✅ Done")
print(f"Repo: https://github.com/{username}/{repo_name}")
print(f"Pages URL after enabling GitHub Pages: https://{username}.github.io/{repo_name}/")
```
